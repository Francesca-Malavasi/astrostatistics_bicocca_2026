```python
import emcee
import numpy as np
import matplotlib.pyplot as plt
import astroML.stats
import astropy
from astropy.visualization import hist
```

```python
N = 10000 #NUMBER OF DAYS

day = 1 #assume the fist day is cloudy 
prediction = []
prediction.append(day)
count_sunny = 0
cumulative_fraction_sunny = []
cumulative_fraction_sunny.append(0) #because the first day is cloudy  

for i in range(1,N): #first step is already given
    if day == 1: #the day is cluody
        day = np.random.choice([0, 1], p=[0.5, 0.5]) #p here in random choise are transition probability 
    else: #so the day is sunny 
        day = np.random.choice([0, 1], p=[0.9, 0.1])

    prediction.append(day)
    if day == 0:
        count_sunny += 1

    cumulative_fraction_sunny.append(count_sunny / (i+1)) 
    
print(count_sunny)
#print(cumulative_fraction_sunny)
```

```python
astropy.visualization.hist(prediction, bins="scott", histtype="step")
plt.xlabel("prediction")
```

```python
x = np.arange(0, N, 1)
print(len(cumulative_fraction_sunny))
print(len(x))
print(x)
plt.plot(x, cumulative_fraction_sunny)
plt.xlabel("x")
plt.ylabel("cumulative fraction of sunny day")
plt.ylim(0.80, 0.86) #this is to study the behaviour of the function in more detail, note the precence of a burn in phase 
```

```python
print("median value of cumulative fraction of sunny day:", np.median(cumulative_fraction_sunny)) #median and median have very similar value
print("mean value of comulative fraction of sunny day:", np.mean(cumulative_fraction_sunny)) 
print("sigma G:", astroML.stats.sigmaG(cumulative_fraction_sunny)) 
print("minimum cumulative fraction:", min(cumulative_fraction_sunny), "max cumulative fraction:", max(cumulative_fraction_sunny))
```

```python
#experimenting different cut-off
burn_in_cutoff = (x > 2000)
burn_in_cutoff_1 = (x > 1000)
burn_in_cutoff_2 = (x > 1500)

mask_sunny = []
mask_sunny_1 = []
mask_sunny_2 = []

for xi,cum in zip(x, cumulative_fraction_sunny):
    if xi > 2000:
        mask_sunny.append(cum)
    if xi > 1000:
        mask_sunny_1.append(cum)
    if xi > 1500:
        mask_sunny_2.append(cum)
       
plt.plot(x[burn_in_cutoff_1], mask_sunny_1, label=r"burn in $x>1000$")
plt.plot(x[burn_in_cutoff_2], mask_sunny_2, label=r"burn in $x>1500$")
plt.plot(x[burn_in_cutoff], mask_sunny, label=r"burn in $x>2000$")#plot taking away the cutoff
plt.xlabel("x")
plt.ylabel("cumulative fraction sunny")
plt.legend()
```

```python
astropy.visualization.hist(mask_sunny_1, bins="freedman", label=r"burn in $x>1000$", alpha =0.5)
astropy.visualization.hist(mask_sunny_2, bins="freedman", label=r"burn in $x>1500$", alpha =0.5)
astropy.visualization.hist(mask_sunny, bins="freedman", label=r"burn in $x>2000$", alpha=0.5)
plt.xlabel("cumulative fraction of sunny day")
```

## stock market finance

```python
N = 10000 #NUMBER OF DAYS

# 0 stands for bull market 
# 1 stands for bear
#2 stands for stagnant

#initial condition 
day = 0 #assume the fist day is bull
prediction = []
prediction.append(day)

#counter
count_bull = 1 #the first is bull
count_bear = 0
count_stagnant = 0

#cumulative fraction 
cumulative_fraction_bull = []
cumulative_fraction_bear = []
cumulative_fraction_stagnant = []

cumulative_fraction_bull.append(1)
cumulative_fraction_bear.append(0) 
cumulative_fraction_stagnant.append(0) 

for i in range(1,N): #first step is already given
    if day == 0: #the day is bull
        day = np.random.choice([0, 1, 2], p=[0.9, 0.075, 0.025]) #p here in random choise are transition probability
    elif day == 1: #the day is bear 
        day = np.random.choice([0, 1, 2], p=[0.15, 0.8, 0.05])
    else:
        day = np.random.choice([0, 1, 2], p=[0.25, 0.25, 0.5])

    prediction.append(day)
    if day == 0:
        count_bull += 1
    elif day == 1: #the day is bear 
        count_bear += 1 
    else:
        count_stagnant += 1 

    cumulative_fraction_bull.append(count_bull / (i+1))
    cumulative_fraction_bear.append(count_bear / (i+1))
    cumulative_fraction_stagnant.append(count_stagnant / (i+1)) 
```

```python
print(count_bull)
print(count_bear)
print(count_stagnant)
print(count_bull+count_bear+count_stagnant)
```

```python
#print(prediction) 
```

```python
x = np.arange(0, N, 1)

print(len(cumulative_fraction_bear))
print(len(cumulative_fraction_bull))
print(len(cumulative_fraction_stagnant))
print(len(x))
print(x)

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8))

ax1.plot(x, cumulative_fraction_bear, color='r')
ax1.set_ylabel("cumulative fraction bear")
ax1.set_xlabel("x")

ax2.plot(x, cumulative_fraction_bull, color='r')
ax2.set_ylabel("cumulative fraction bull")
ax2.set_xlabel("x")

ax3.plot(x, cumulative_fraction_stagnant, color='r')
ax3.set_ylabel("cumulative fraction stagnant") 
ax3.set_xlabel("x")

plt.subplots_adjust(hspace=0.5)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8))

ax1.hist(cumulative_fraction_bear, bins=50, color='r', alpha=0.5)
ax1.set_xlabel("cumulative fraction bear")

ax2.hist(cumulative_fraction_bull, bins=50, color='r', alpha=0.5)
ax2.set_xlabel("cumulative fraction bull")

ax3.hist(cumulative_fraction_stagnant, bins=50, color='r', alpha=0.5)
ax3.set_xlabel("cumulative fraction stagnant") 

plt.subplots_adjust(hspace=0.5)
```

```python
burn_in_cutoff = (x > 2000)

mask_bear = []
mask_bull = []
mask_stagnant = []

for xi,cum_be, cum_bu, cum_s in zip(x, cumulative_fraction_bear, cumulative_fraction_bull, cumulative_fraction_stagnant):
    if xi > 2000:
        mask_bear.append(cum_be)
        mask_bull.append(cum_bu)
        mask_stagnant.append(cum_s)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8))

ax1.plot(x, cumulative_fraction_bear, color='r')
ax1.plot(x[burn_in_cutoff],mask_bear, label="cutting burn-in", color="blue")
ax1.set_ylabel("cumulative fraction bear")
ax1.set_xlabel("x")

ax2.plot(x, cumulative_fraction_bull, color='r')
ax2.plot(x[burn_in_cutoff],mask_bull, label="cutting burn-in", color="blue")
ax2.set_ylabel("cumulative fraction bull")
ax2.set_xlabel("x")

ax3.plot(x, cumulative_fraction_stagnant, color='r')
ax3.set_ylabel("cumulative fraction stagnant")
ax3.plot(x[burn_in_cutoff],mask_stagnant, label="cutting burn-in", color="blue")
ax3.set_xlabel("x")

plt.subplots_adjust(hspace=0.5)
plt.legend()
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 8))

ax1.hist(cumulative_fraction_bear, color='r', alpha=0.5, density=True)
ax1.hist(mask_bear, color='blue', alpha=0.5, label="burn-in sample", density=True)
ax1.set_xlabel("cumulative fraction bear")

ax2.hist(cumulative_fraction_bull, color='r', alpha=0.5, density=True)
ax2.hist(mask_bull, color='blue', alpha=0.5, label="burn-in sample", density=True)
ax2.set_xlabel("cumulative fraction bull")

ax3.hist(cumulative_fraction_stagnant, color='r', alpha=0.5, density=True)
ax3.hist(mask_stagnant, color='blue', alpha=0.5, label="burn-in sample", density=True)
ax3.set_xlabel("cumulative fraction stagnant") 

plt.subplots_adjust(hspace=0.5)
plt.legend()
```
